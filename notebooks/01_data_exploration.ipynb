# Data Exploration: CBIS-DDSM

This notebook performs a class-distribution analysis and verifies the ROI masks as per Chunk 0 of the Implementation Plan.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from PIL import Image

# Set seaborn style
sns.set_theme(style="whitegrid")
```

## 1. Load the Manifest

```python
MANIFEST_PATH = '../data/manifest.csv'
df = pd.read_csv(MANIFEST_PATH)
df.head()
```

## 2. Class Distribution Analysis

We need to document the imbalance between Benign and Malignant cases, and also between Mass and Calcification abnormality types.

```python
plt.figure(figsize=(10, 5))

# Benign vs Malignant (after merging BENIGN_WITHOUT_CALLBACK)
plt.subplot(1, 2, 1)
sns.countplot(data=df, x='label', palette='Set2')
plt.title('Class Distribution (Benign vs Malignant)')
plt.ylabel('Count')

# Mass vs Calcification
plt.subplot(1, 2, 2)
sns.countplot(data=df, x='abnormality_type', palette='Set3')
plt.title('Abnormality Type (Mass vs Calcification)')
plt.ylabel('Count')

plt.tight_layout()
plt.show()

print("Value Counts - Label:")
print(df['label'].value_counts(normalize=True) * 100)
print("\nValue Counts - Abnormality Type:")
print(df['abnormality_type'].value_counts(normalize=True) * 100)
```

## 3. Verify ROI Masks
Let's load a sample image and its corresponding ROI mask to verify alignment.

```python
RAW_DATA_DIR = '../data/raw'

# Pick a random sample
sample = df.iloc[0]
img_path = os.path.join(RAW_DATA_DIR, sample['image_path'])
mask_path = os.path.join(RAW_DATA_DIR, sample['roi_mask_path'])

try:
    img = Image.open(img_path)
    mask = Image.open(mask_path)
    
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 3, 1)
    plt.imshow(img, cmap='gray')
    plt.title(f"Image: {sample['label']}")
    plt.axis('off')
    
    plt.subplot(1, 3, 2)
    plt.imshow(mask, cmap='gray')
    plt.title("ROI Mask")
    plt.axis('off')
    
    plt.subplot(1, 3, 3)
    plt.imshow(img, cmap='gray')
    plt.imshow(mask, cmap='jet', alpha=0.4) # Overlay
    plt.title("Overlay")
    plt.axis('off')
    
    plt.tight_layout()
    plt.show()
except Exception as e:
    print(f"Error loading images: {e}")
```
